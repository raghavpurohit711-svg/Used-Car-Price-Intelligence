```python
import pandas as pd

df = pd.read_csv(r"C:\Programming\Machine Learning\Used Car Price Intelligence\Used Car Arbitrage\data\raw\raw_car_listings.csv")
df.head()
```

```python
import re

def extract_final_price(price_str):

    if pd.isna(price_str):
        return None

    numbers = re.findall(r"\d+\.\d+|\d+", str(price_str))

    if numbers:
        final_price_lakh = float(numbers[-1])

        return int(final_price_lakh*100000)

    return None

df['Clean_Price'] = df['Price'].apply(extract_final_price)

df[['Price', 'Clean_Price']].head()
```

```python
print(f"Total rows processed: {df.shape[0]}")

df[['Price','Clean_Price']].sample(10)
```

```python
def clean_kilometer(km_str):

    if pd.isna(km_str):
        return None

    clean_str = str(km_str).replace(',','').replace(' km','').replace('km','').strip()

    numbers = re.findall(r'\d+', clean_str)

    if numbers:
        return int(''.join(numbers))

    return None 

df['Clean_Kilometers'] = df['Kilometers'].apply(clean_kilometer)

df[['Kilometers', 'Clean_Kilometers']].sample(5)
```

```python
df['Title'] = df['Title'].fillna('')

title_splits = df['Title'].str.split(' ', n = 2, expand = True)

df['Year'] = title_splits[0]
df['Brand'] = title_splits[1]
df['Model'] = title_splits[2]

df['Year'] = pd.to_numeric(df['Year'],errors='coerce')

df_clean = df.drop(columns=['Title', 'Price', 'Kilometers'])
df_clean.head()
```

```python
df_clean.to_csv(r"C:\Programming\Machine Learning\Used Car Price Intelligence\Used Car Arbitrage\data\processed\clean_car_listing.csv", index = False)
print("Clean data Successfully saved to data/processed")
```
